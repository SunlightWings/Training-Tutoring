### Assignment 1: Guessing Game:

* Create a numbers guessing game using loops
* Print too high or too low if user guesses incorrectly.

**Q. Whats the best way to play the game to maximize winning chance?**

```python
## Code for Guessing game:

import random

def guess_number():
  print("____Guessing Game____")
  print("Thinking of a number between 1 and 100")

  number_to_guess = random.randint(1, 100)        # random_number generate vayo
  attempts = 0                                    # 1. initialization
  max_attempts = 10

  while attempts < max_attempts:                 # 2. looping till when (condition)
    guess = int(input("Enter your guess: "))
    attempts += 1                                # 3. counter decrement/increment

    if guess < number_to_guess:
      print("Too low! Try again.")
    elif guess > number_to_guess:
      print("Too high! Try again.")
    else:
      print(f"Congratulations! You guessed the number {number_to_guess} in {attempts} attempts.")
      break
```

```python
guess_number()
```

### Assignment 2: Guessing Game (again):

Use data structures this time.
* Set to keep track of already guessed numbers (remember no duplicates in set)
* List to keep track of all the guesses
* Dictionary to give hints by just storing if its high or low.
* Tuple to create a concrete immutable group of numbers to choose from.

```python
empty = set(['1'])
empty
```

```python
## Guessing game code here:

import random

def guess_number():
  number_range = (1,100)

  # dictionary: stores hints (too high, too low)
  hints = {'high': 'Too high', 'low': 'Too low!!'}

  # set (duplicates wont be printed)
  already_numbers = set()

  # list (players haru ko non/duplicate guesses)
  past_guesses = []

  target_number = random.randint(number_range[0], number_range[1])     # actual number, genrated by computer (answer key)

  print("____Guessing Game____")
  print("Think of a number between 1 and 100")

  while True:
    guess = int(input("Enter your guess: "))

    if guess in already_numbers:
      print("You already guessed that number. Try again.")
      continue

    already_numbers.add(guess)
    past_guesses.append(guess)

    if guess < target_number:
      print(hints['low'])
    elif guess > target_number:
      print(hints['high'])
    else:
      print(f"Congratulations! You guessed the number {target_number} in {len(past_guesses)} attempts.")
      break


```

```python
guess_number()
```

## FIle handling:

1. Text file: reading and writing: (try on Vs code)


```python
# Writing to a text file
def write_to_file(filename, text):
    with open(filename, 'w') as file:
        file.write(text)
    print(f"Text written to {filename}")

# Reading from a text file
def read_from_file(filename):
    try:
        with open(filename, 'r') as file:
            content = file.read()
        print(f"Content of {filename}: \n{content}")
    except FileNotFoundError:
        print(f"{filename} not found.")

# Example usage
filename = "example.txt"
text = "Hello, this is a sample text file.\nFile handling in Python is easy!"
write_to_file(filename, text)
read_from_file(filename)
```
